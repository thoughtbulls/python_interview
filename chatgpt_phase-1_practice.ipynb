Python Q001 ⭐⭐☆☆☆ (Interview Warm-up)
Business Scenario

You are processing application logs.

A log processing job needs to identify the first event ID that appears only once in a stream of events.

This is useful for detecting unique events such as a one-time login or a failed transaction that occurred only once.

Input:
* A string consisting of lowercase English letters.
* Example: "swiss"

Return:
* Return the first non-repeating character.
* If every character repeats, return: None

```python
def first_non_repeating_char(s: str):
    _counts = {}

    for c in s:
        if c not in _counts:
            _counts[c] = 1
        else:
            _counts[c] +=1

    for c in s:
        if _counts[c] == 1:
            return c


if __name__ == "__main__":
    
    fc = first_non_repeating_char("swiss")
    print(fc)
```

Return a dictionary where:

Expected output:
* {

    * "HR": ["Bob", "Eva"],
    * "Finance": ["David"]

**Constraints**:

```python
employees = [
    {"emp_id": 1, "name": "Alice", "dept": "IT"},
    {"emp_id": 2, "name": "Bob", "dept": "HR"},
    {"emp_id": 3, "name": "Charlie", "dept": "IT"},
    {"emp_id": 4, "name": "David", "dept": "Finance"},
    {"emp_id": 5, "name": "Eva", "dept": "HR"},
]

def group_by_department(employees):
    depts = {}

    for emp in employees:
        if emp['dept'] not in depts:
            depts[emp['dept']] = [emp]
        else:
            depts[emp['dept']].append(emp)
    
    return depts

if __name__ == "__main__":
    result = group_by_department(employees)

    print(result)
```

```python
transactions = [ 
                {"customer": "Alice", "amount": 100}, 
                {"customer": "Bob", "amount": 200}, 
                {"customer": "Alice", "amount": 50}, 
                {"customer": "David", "amount": 300}, 
                {"customer": "Bob", "amount": 150}, 
                {"customer": "Alice", "amount": 25}
]

def customer_total_spend(transactions):
    customer_total = {}

    for transaction in transactions:
        customer = transaction['customer']
        amount = transaction['amount']

        stats = customer_total.setdefault(customer, {"total": 0, "orders": 0, "average": 0.0})
       
        stats['total'] += amount
        stats['orders'] += 1
    
    for c in customer_total:
        stats = customer_total[c]
        stats['average'] = round(stats['total'] / stats['orders'], 2)
    
    return customer_total


if __name__ == "__main__":
    result = customer_total_spend(transactions)
    print(result)
```

**Interview Question Q006 ⭐⭐⭐☆☆**
**Business Scenario**

A file has been loaded into memory.

Each row represents a transaction.

The Finance team wants to identify **duplicate transaction IDs**.

Input:
*transactions = [
    {"txn_id": 101, "amount": 100},
    {"txn_id": 102, "amount": 200},
    {"txn_id": 101, "amount": 150},
    {"txn_id": 103, "amount": 400},
    {"txn_id": 102, "amount": 500},
]

Return:
*Return the row numbers (1-based) where duplicate transaction IDs occur.

Expected output
* {
    101: [3],
    102: [5]
}

```python
transactions = [
    {"txn_id": 101, "amount": 100},
    {"txn_id": 102, "amount": 200},
    {"txn_id": 101, "amount": 150},
    {"txn_id": 103, "amount": 400},
    {"txn_id": 102, "amount": 500},
    {"txn_id": 101, "amount": 120},
]


def duplicate_transactions(transactions):
    seen = set()
    duplicates = {}

    for row, txn in enumerate(transactions, start=1):
        txn_id = txn["txn_id"]

        if txn_id not in seen:
            duplicates.setdefault(txn_id, []).append(row)
        else:
            seen.add(txn_id)

    return duplicates


if __name__ == "__main__":
    result = duplicate_transactions(transactions)
    print(result)
```

**Q011 ⭐⭐⭐⭐☆ (A Real Data Engineering Problem)**

This one is inspired by ETL validation.

**Business Scenario**

Yesterday's customer file and today's customer file have arrived.

The business wants to know:

1. **New customers**(present today but not yesterday)
2. **Customers who left** (present yesterday but not today)
3. **Customers who remained** (present in both)

**Input**:
* yesterday = [101, 102, 103, 104, 105, 106]
* today = [103, 104, 105, 106, 107, 108, 109]

**Expected Output**:

* {
*     "new": [107, 108, 109],
*     "left": [101, 102],
*     "existing": [103, 104, 105, 106]
* }

**Constraints**:
* Use set operations only.
* No loops.
* Return sorted lists.
* Keep the code concise and readable.

This is a genuine ETL reconciliation task and will reinforce all the set operations you've just learned.

```python
yesterday = [101, 102, 103, 104, 105, 106]
today = [103, 104, 105, 106, 107, 108, 109]

def customer_changes(yesterday, today):
    set_yesterday = set(yesterday)
    set_today = set(today)

    # New customers
    new_customers = set_today.difference(set_yesterday)

    # Left customers
    left_customers = set_yesterday.difference(set_today)

    # Customers who remained
    remained_customers = set_today.intersection(set_yesterday)

    return {
        "new": sorted(new_customers),
        "left": sorted(left_customers),
        "existing": sorted(remained_customers)
    }

if __name__ == "__main__":
    result = customer_changes(yesterday, today)
    print(result['new'])
```

```python
logs = [
    "ERROR",
    "INFO",
    "ERROR",
    "WARNING",
    "INFO",
    "ERROR",
    "DEBUG",
    "WARNING",
    "INFO"
]

def log_level_count(logs):
    log_counts = {}
    for log in logs:
        log_counts[log] = log_counts.get(log, 0) + 1
    
    # Follow-up 2
    mx = max(log_counts.values())
    log_max_count = [(log, log_counts[log]) for log in log_counts if log_counts[log] == mx]

    return log_max_count


if __name__ == "__main__":
    result = log_level_count(logs)
    print(result)
```

**Q019 ⭐⭐⭐⭐☆ (Exception Handling)**
**Business Scenario**

Your ETL pipeline receives transaction records from an API.

Unfortunately, some records are malformed:
- Missing amount
- Amount stored as a string
- Invalid numeric values

Your job is to calculate the **total transaction amount**.

If a record is invalid, **skip it** and continue processing.

**Input:**
- transactions = [
-     {"txn_id": 1, "amount": "100.50"},
-     {"txn_id": 2, "amount": 250},
-     {"txn_id": 3},                    # Missing amount
-     {"txn_id": 4, "amount": "abc"},   # Invalid number
-     {"txn_id": 5, "amount": 99.99},
- ]

**Expected Output:**
450.49

Because
- 100.50
- +250.00
- +99.99
- -------
- 450.49

The other two records should simply be ignored.

**Constraints**:
- One pass.
- Do not stop processing if one record is bad.
- Return the total as a float.

```python
transactions = [
    {"txn_id": 1, "amount": "100.50"},
    {"txn_id": 2, "amount": 250},
    {"txn_id": 3},                    # Missing amount
    {"txn_id": 4, "amount": "abc"},   # Invalid number
    {"txn_id": 5, "amount": 99.99},
    {"txn_id": 6, "amount": None}
]

def total_amount(transactions):
    total = 0.0
    skipped = []
    for txn in transactions:
        try:
            total = total + float(txn["amount"])
        except (KeyError, ValueError, TypeError):
            skipped.append(txn['txn_id'])
    
    return (total, skipped)  


if __name__ == "__main__":
    result = total_amount(transactions)
    print(result)
```

```python
customers = [
    {
        "customer_id": 101,
        "name": "Rahul",
        "email": "rahul@test.com",
        "city": "Delhi",
        "active": True
    },
    {
        "customer_id": 102,
        "name": "Priya",
        "city": "Mumbai",
        "active": True
    },
    {
        "customer_id": 103,
        "name": "Amit",
        "email": "",
        "city": "Bangalore",
        "active": False
    },
    {
        "customer_id": 104,
        "name": "Sneha",
        "email": "sneha@test.com",
        "city": "Hyderabad",
        "active": True
    },
    {
        "customer_id": 105,
        "name": "Vikas",
        "email": None,
        "city": "Lucknow",
        "active": True
    },
    {
    "customer_id":106,
    "name":"John",
    "email":"john@test.com",
    "active":True
    }
]

# from collections import defaultdict

def clean_customers(customers):
    clean_customers = {}
    skipped_customer_ids = []
    for customer in customers:
        try:
            if not customer['active']:
                raise ValueError

            email = customer['email']
            if email is None or email.strip() == "":
                raise ValueError
            
            city = customer.get("city", "Unknown")
            
            clean_customers[customer['customer_id']] = {
                        'name': customer['name'],
                        'email': email,
                        'city': city
                    }
        except (KeyError, ValueError):
            skipped_customer_ids.append(customer['customer_id'])

    return clean_customers, skipped_customer_ids



if __name__ == "__main__":
    result = clean_customers(customers)
    print(result)
```

```python
# Method 2: With setdefault()
category_map = {}
for sale in sales:
    category = sale['category']
    product_id = sale['product_id']

    category_map.setdefault(category, []).append(product_id)

print(category_map)
```
